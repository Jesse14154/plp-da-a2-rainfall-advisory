## 1. Introduction

This project analyzes monthly rainfall data for five Kenyan counties: Nairobi, Nakuru, Kisumu, Mombasa, and Machakos.

The analysis uses Python, NumPy, and the CSV module to clean rainfall data, calculate rainfall statistics, identify wet and dry periods, and recommend suitable planting windows.

The final output is a rainfall summary and a planting advisory to support agricultural planning.

```python
import numpy as np
import csv
```

```python
# Raw Machakos rainfall data for January to December
machakos_raw = [
    "48", "35", "88", "N/A", "140", " 20 ",
    "9", "12", "", "45", "182", "96"
]

# Function to clean rainfall values
def clean_rainfall(value):
    try:
        return int(value.strip())
    except ValueError:
        return None

# Convert raw values into clean values
machakos_clean = [clean_rainfall(value) for value in machakos_raw]

# Identify valid rainfall values
valid_values = [value for value in machakos_clean if value is not None]

# Calculate the average of valid rainfall values
average_rainfall = round(sum(valid_values) / len(valid_values))

# Replace invalid values with the average
machakos_clean = [
    value if value is not None else average_rainfall
    for value in machakos_clean
]

# Identify the months that were imputed
months = [
    "January", "February", "March", "April",
    "May", "June", "July", "August",
    "September", "October", "November", "December"
]

imputed_months = [
    months[i] for i, value in enumerate(machakos_raw)
    if clean_rainfall(value) is None
]

# Display the results
print("Clean Machakos Rainfall Data:")
print(machakos_clean)

print("\nAverage rainfall used for imputation:")
print(average_rainfall, "mm")

print("\nMonths with imputed values:")
print(imputed_months)
```

```python
# County names
counties = ["Nairobi", "Nakuru", "Kisumu", "Mombasa", "Machakos"]

# Monthly rainfall data for the first four counties
clean_rows = [
    [64, 56, 92, 219, 176, 35, 18, 24, 31, 61, 165, 104],  # Nairobi
    [35, 42, 70, 145, 120, 78, 96, 105, 72, 55, 60, 38],   # Nakuru
    [58, 90, 150, 240, 180, 95, 70, 85, 90, 88, 130, 95],  # Kisumu
    [30, 15, 55, 150, 240, 110, 90, 65, 60, 90, 100, 70]   # Mombasa
]

# Combine all five counties
all_rainfall = clean_rows + [machakos_clean]

# Convert the dataset into a NumPy array
rainfall = np.array(all_rainfall)

# Display the array information
print("County names:")
print(counties)

print("\nRainfall data:")
print(rainfall)

print("\nArray shape:")
print(rainfall.shape)

print("\nArray data type:")
print(rainfall.dtype)
```

```python
# Calculate annual rainfall totals for each county
annual_totals = rainfall.sum(axis=1)

# Calculate average rainfall for each month across all counties
monthly_averages = rainfall.mean(axis=0)

# Display annual rainfall totals
print("ANNUAL RAINFALL TOTALS")
for county, total in zip(counties, annual_totals):
    print(f"{county}: {total} mm")

# Display monthly averages
months = [
    "January", "February", "March", "April",
    "May", "June", "July", "August",
    "September", "October", "November", "December"
]

print("\nMONTHLY AVERAGE RAINFALL")
for month, average in zip(months, monthly_averages):
    print(f"{month}: {average:.2f} mm")
```

```python
# Find the index of the wettest month for each county
wettest_month_indices = np.argmax(rainfall, axis=1)

# Display the wettest month for each county
print("WETTEST MONTH FOR EACH COUNTY")

for county, index in zip(counties, wettest_month_indices):
    print(f"{county}: {months[index]}")
```

```python
# Identify months with rainfall below 50 mm
dry_month_mask = rainfall < 50

# Count dry months for each county
dry_month_counts = dry_month_mask.sum(axis=1)

# Display dry month counts
print("DRY MONTHS PER COUNTY (Below 50 mm)")

for county, count in zip(counties, dry_month_counts):
    print(f"{county}: {count} dry months")
```

```python
# Calculate combined rainfall for every pair of consecutive months
consecutive_rainfall = rainfall[:, :-1] + rainfall[:, 1:]

# Find the position of the highest combined rainfall for each county
best_window_indices = np.argmax(consecutive_rainfall, axis=1)

# Calculate the highest combined rainfall for each county
best_window_totals = np.max(consecutive_rainfall, axis=1)

# Display the best planting windows
print("BEST TWO-MONTH PLANTING WINDOWS")

for county, index, total in zip(
    counties, best_window_indices, best_window_totals
):
    first_month = months[index]
    second_month = months[index + 1]

    print(f"{county}: {first_month} - {second_month} ({total} mm)")
```

```python
# Define the CSV column headings
fieldnames = [
    "county",
    "annual_total_mm",
    "wettest_month",
    "dry_months",
    "planting_window"
]

# Create the rainfall summary CSV file
with open("rainfall_summary.csv", "w", newline="") as file:

    writer = csv.DictWriter(file, fieldnames=fieldnames)

    # Write the column headings
    writer.writeheader()

    # Write one summary row for each county
    for county, total, wet_index, dry_count, window_index in zip(
        counties,
        annual_totals,
        wettest_month_indices,
        dry_month_counts,
        best_window_indices
    ):

        writer.writerow({
            "county": county,
            "annual_total_mm": int(total),
            "wettest_month": months[wet_index],
            "dry_months": int(dry_count),
            "planting_window": (
                f"{months[window_index]}-{months[window_index + 1]}"
            )
        })

print("rainfall_summary.csv created successfully!")
```

```python
# Read and display the generated CSV file
with open("rainfall_summary.csv", "r") as file:
    print(file.read())
```

## Planting Advisory

Based on the rainfall analysis, the NGO should consider April–May as the main rainfall-based planting window for Nairobi, Nakuru, Kisumu, and Mombasa, as these counties receive their highest combined rainfall during these consecutive months.

Machakos has its highest combined rainfall in November–December, making this a potential alternative planting window for the county.

Kisumu recorded the highest annual rainfall total at 1,371 mm, while Machakos recorded the lowest at 811 mm.

Machakos also recorded six dry months, indicating a greater need for careful water management and drought preparedness. Nakuru recorded three dry months, while Nairobi and Mombasa recorded four and two respectively.

The NGO should use these findings to guide local agricultural planning while considering crop water requirements, soil conditions, and advice from agricultural extension officers.

## Data Quality Notes

The original Machakos rainfall dataset contained two invalid entries: "N/A" for April and an empty value for September.

These invalid entries were handled using a Python cleaning function with `try` and `except ValueError`. The average of the ten valid Machakos rainfall values was 67.5 mm, which was rounded to 68 mm and used to replace both invalid entries.

After cleaning, the rainfall data for all five counties was combined into a NumPy array with a shape of (5, 12) and an integer data type.

The analysis assumes that the supplied rainfall measurements are accurate apart from the identified Machakos entries. Since the dataset covers only one year, the identified planting windows should not be treated as guaranteed seasonal patterns. Longer-term rainfall records and local agricultural information are recommended before making final planting decisions.
